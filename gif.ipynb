```python
import os
from PIL import Image

def convert_gifs_to_webp(root_dir):
    """
    Converts all GIF files in a directory and its subdirectories to WebP format
    with the highest quality settings.

    Args:
        root_dir: The root directory to search for GIF files.
    """
    for root, _, files in os.walk(root_dir):
        for file in files:
            if file.lower().endswith('.gif'):
                gif_path = os.path.join(root, file)
                webp_path = os.path.splitext(gif_path)[0] + ".webp"

                try:
                    # Open the GIF
                    img = Image.open(gif_path)

                    # Extract GIF frames and duration
                    frames = []
                    durations = []
                    for i in range(img.n_frames):
                        img.seek(i)
                        frames.append(img.copy())
                        durations.append(img.info['duration'])

                    # Save as WebP with highest quality (lossless=True for animation)
                    frames[0].save(webp_path, save_all=True, append_images=frames[1:], 
                                   lossless=True, method=6, quality=100, 
                                   duration=durations, loop=0, disposal=2)

                    print(f"Converted {gif_path} to {webp_path}")

                except Exception as e:
                    print(f"Error converting {gif_path}: {e}")

if __name__ == "__main__":
    root_directory = "."  # You can change this to your desired directory
    convert_gifs_to_webp(root_directory)
```
